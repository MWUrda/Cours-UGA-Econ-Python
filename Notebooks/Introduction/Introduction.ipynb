### <center> 3. Exemple d'un budget mensuel </center>

Supposons qu'un ménage dispose d'un revenu mensuel et effectue quatre types de dépenses. Les chiffres ci-dessous sont fictifs et exprimés en euros.

L'équation comptable est la suivante :

$$
\text{argent restant} = \text{revenu} - \text{dépenses totales}.
$$


Si le solde est positif, la somme peut être épargnée. S'il est négatif, les dépenses excèdent le revenu ; le ménage  doit alors réduire ses dépenses, puiser dans son épargne ou emprunter.

```python
revenu_mensuel = 3000
loyer = 800
alimentation = 800
transport = 200
autres_dépenses = 800

dépenses_totales = loyer + alimentation + transport + autres_dépenses
argent_restant = revenu_mensuel - dépenses_totales

print("Dépenses totales :", dépenses_totales, "€")
print("Argent restant :", argent_restant, "€")
```

#### Lisez le code ligne par ligne

- `revenu_mensuel = 3000` enregistre la valeur `12000` sous le nom `revenu_mensuel`.
- En Python, `=` signifie **enregistrer la valeur située à droite sous le nom situé à gauche**.
- Des noms explicites comme `revenu_mensuel` rendent le code plus facile à comprendre que des noms tels que `x`.
- `dépenses_totales = loyer + alimentation + transport + autres_dépenses` additionne les quatre catégories de dépenses.
- `argent_restant = revenu_mensuel - dépenses_totales` applique l'équation budgétaire.
- Les deux lignes `print` affichent les résultats accompagnés de courtes étiquettes.

Python ne connaît pas l'économie par lui-même. Nous traduisons la relation économique en étapes claires que Python peut calculer.

> **Essayez vous-même :** Le code ci-dessous utilise le même budget. Modifiez uniquement `loyer_exercice`. Estimez le loyer requis pour avoir une somme restante de 500 €.


```python
loyer_exercise = 800  # modifiez ce nombre de manière à avoir une somme restante de 500 €.

dépenses_exercise = loyer_exercise + alimentation + transport + autres_dépenses
argent_restant = revenu_mensuel - dépenses_exercise

print("Argent restant :",argent_restant, "€")
```

### <center> 4. Laisser le code prendre une décision </center>

Une instruction `if` permet à Python de choisir entre plusieurs options.


Ici, la règle économique est simple :


- si le solde est nul ou positif, le budget est équilibré ou excédentaire ;
- sinon, le budget est déficitaire.

```python
test_argent_restant = -500

if test_argent_restant >= 0 :
    
    print("Le budget est équilibré ou excédentaire.")

else:
    print("Le budget est déficitaire.")
```

Étudiez attentivement les nouveaux symboles :

- `if` introduit le premier cas.
- `>=` signifie **supérieur ou égal à**.
- Le deux-points `:` indique à Python qu'un bloc en retrait suit.
- Les quatre espaces avant `print` indiquent que cette ligne appartient au cas `if`.
- `else` définit l'alternative lorsque la condition est fausse.

### <center> 5. Données réelles sur l'inflation : trois façons de les obtenir </center>

Nous passons maintenant de chiffres fictifs à des données économiques publiées.

Nous utilisons les données sur l'[Indice des Prix à la Consommation](https://www.insee.fr/fr/statistiques/4268033#onglet-1) fournies par l'INSEE et utilisées comme mesure de l'inflation en France.

Nous allons obtenir ces données selon deux méthodes, la deuxième plus pratique et générale que la première :

1. copier manuellement quelques observations depuis le site web,
2. lire les observations à partir d'un fichier CSV.

#### Que signifie le taux d'inflation ?

L'indice des prix à la consommation synthétise les prix d'un large panier de biens et de services achetés par les ménages. Le taux d'inflation mesure la vitesse à laquelle ce niveau général des prix évolue.

Un taux d'inflation annuel de 8,5 % en 2022 signifie que le niveau moyen des prix était supérieur de 8,5 % à celui de 2021.

Cela ne signifie **pas** que :

- tous les prix ont augmenté exactement de 8,5 %, ou
- les prix ont baissé lorsque l'inflation a par la suite diminué.

Si l'inflation passe de 8,5 % à 3,4 %, les prix continuent d'augmenter, mais plus lentement. Les prix ne baissent en moyenne que lorsque l'inflation est négative. C'est ce qu'on appelle la déflation.

Ces données nous indiquent **ce qui est arrivé au niveau des prix**.

#### Méthode 1 : Copier quelques observations à la main

1. Ouvrez la [page de sur l'Indice des Prix à la Consommation de l'INSEE](https://www.insee.fr/fr/statistiques/4268033#onglet-1)
2. Repérez la liste des observations sous le graphique et sélectionnez **View All** (Tout afficher) si nécessaire.
3. Identifiez les observations pour les années 2021, 2022, 2023, 2024, et 2025.
4. Saisissez chaque valeur dans la liste Python ci-dessous.
5. Conservez le même ordre pour les années et les taux d'inflation.

Ces cinq observations peuvent encore être vérifiées visuellement en choisissant **Graphique** à la place de **Tableau** mais il est évident qu'avec une longue série chronologique, la copie manuelle devient fastidieuse et les erreurs deviennent potentiellement fréquentes.

```python
annees = [2021, 2022, 2023, 2024, 2025]
# 'annees' est une variable. La valeur qu'elle contient est une liste. Cette liste comprend quatre entiers.
inflation_taux = [
    1.6,  # 2021
    5.2,  # 2022
    4.9,  # 2023
    2.0,  # 2024
    0.9  # 2025
]
# inflation_taux  est une autre variable. Elle contient une liste de quatre nombres à virgule flottante (c'est-à-dire des nombres décimaux).
print("Années:", annees) # print est une fonction
print("Taux d'inflation:", inflation_taux)
```

Ce sont deux listes associées l'une à l'autre:

- la première année, `2021`, correspond au premier taux d'inflation, `1,6` ;
- la deuxième année, `2022`, correspond au deuxième taux d'inflation, `5.2` ;
- et ainsi de suite.

Cet ordre fait partie intégrante des données. Si l'ordre est incorrect, le résultat le sera également.

Nous pouvons également calculer le taux d'inflation annuel moyen sur ces quatre années :

```python
inflation_moyenne = sum(inflation_taux) / len(inflation_taux)

print("Inflation Annuelle Moyenne:", inflation_moyenne, "pourcentage")
```

La moyenne arithmétique résume ces cinq taux annuels. Elle ne représente pas la hausse totale du niveau des prix entre 2021 et 2025. Pour déterminer cette hausse totale, il faudrait partir d'un prix donné et appliquer successivement les augmentations de chaque année.

Avant d'examiner le graphique, répondez aux questions suivantes :

1. Au cours de quelle année l'inflation a-t-elle été la plus élevée ?
2. La baisse de l'inflation après 2022 a-t-elle entraîné une diminution du niveau moyen des prix ?

```python
import matplotlib.pyplot as plt # nous avons besoin de cette bibliothèque pour créer des graphiques ; vous la verrez dans la suite du cours,
# elle ne fait pas partie de la bibliothèque standard de Python, nous devons donc d'abord l'importer

plt.plot(annees, inflation_taux) # nous voulons tracer ces deux listes l'une par rapport à l'autre : la première correspond à l'axe des x et la seconde à l'axe des y
plt.xlabel("Année") 
plt.ylabel("Taux d'inflation (pourcentage)")
plt.title("Inflation annuelle en France")
plt.show() # cette commande affiche le graphique ; si vous ne l'exécutez pas, le graphique ne s'affichera pas dans le notebook
```

#### Méthode 2 : Lire les données à partir d'un fichier CSV

Saisir des valeurs en un nombre peu élevé comme précédemment demeure une tâche gérable du point de vue de l'attention portée aux erreurs de saisie. En revanche, plus on a des valeurs à saisir et plus le nombre d'erreurs potentielles de saisie devient grand et l'attention qui leur est portée plus difficile.

Une alternative consiste à éviter la saisie manuelles des données et à la place de celle-ci d'importer directement un fichier de données qui les contient(quand celui-ci est disponible) 

Dans le cas présent ces fichier est de type **xlsx**, c.à.d. le format de fichier standard pour les feuilles de calcul de Microsoft Excel. Ce fichier est disponible à l'adresse(sous l'icône **Données** ): https://www.insee.fr/fr/statistiques/fichier/4268033/Essentiel_Inflation_donnees.xlsx

Le code contenu dans la cellule suivante permet d'importer ces données.


```python
# %pip install openpyxl 
# pandas needs this library installed to read Excel files
```

```python
import pandas as pd #On utilise la bibliothèque pandas pour lire des données à partir d'un fichier xlsx. 
                    # Notez que pandas est une bibliothèque d'analyse de données pour Python.

file = "https://www.insee.fr/fr/statistiques/fichier/4268033/Essentiel_Inflation_donnees.xlsx"
inflation_df = pd.read_excel(file,
    sheet_name='Figure 1', skiprows = 3, skipfooter= 3, header = 0)
inflation_df # On affiche les données dans le notebook ; 
             # vous pouvez également utiliser `inflation_data.head()` pour voir les premières lignes des données.
```

Lisez le code ligne par ligne :

- `import pandas as pd` charge le package utilisé pour les tableaux et lui attribue le nom abrégé `pd`.
- `pd.read_excel(...)` lit le fichier xlsx.
- `inflation_data = ...` enregistre le tableau importé sous le nom `inflation_data`.
- Écrire `inflation_data` sur la dernière ligne permet d'afficher le tableau.

Les données sont désormais disponibles dans Python sans avoir à saisir chaque observation dans le notebook.

On remarque que la colonne qui contient le taux d'inflation porte le nom "Indice des prix à la consommation" que nous pouvons 
changer en taux d'inflation afin d'utiliser le même nom que l'on avait utilisé avec la première méthode. Nous allons aussi pour la même raison changer le nom de la colonne qui contient les années.

```python
inflation_df = inflation_df.rename(columns={'Année ' : 'annees', 'Indice des prix à la consommation': 'inflation_taux'})
inflation_df
```

Le code suivant nous permet de: 
- calculer l'inflation annuelle moyenne,
- tracer un graphique du taux d'inflation selon les années.

```python
moyenne_inflation_df = inflation_df["inflation_taux"].mean()


print("Inflation Annuelle Moyenne:", moyenne_inflation_df, "pourcentage")


plt.plot(inflation_df['annees'], inflation_df['inflation_taux'])
plt.xlabel('Année')
plt.ylabel("Taux d'Inflation (pourcentage)")
plt.title('Inflation annuelle en France')
plt.show()
```
